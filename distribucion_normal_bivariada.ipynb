# Distribución Normal Bivariada

## Introducción

La distribución normal bivariada es una extensión de la distribución normal univariada para dos variables aleatorias. Es muy útil cuando queremos estudiar la relación entre dos variables que siguen una distribución normal conjunta.

### Parámetros principales:

- **μ₁, μ₂**: Medias de las variables X e Y
- **σ₁, σ₂**: Desviaciones estándar de X e Y
- **ρ (rho)**: Coeficiente de correlación entre X e Y (donde -1 ≤ ρ ≤ 1)

La función de densidad tiene la forma de una "campana" en 3D, y cuando ρ ≠ 0, las variables están correlacionadas.

```python
# Importamos las librerías necesarias
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import multivariate_normal
from mpl_toolkits.mplot3d import Axes3D
import seaborn as sns

# Configuración para gráficos más bonitos
plt.style.use('seaborn-v0_8-darkgrid')
sns.set_palette("husl")
```

## Ejemplo 1: Distribución Normal Bivariada sin correlación (ρ = 0)

Cuando ρ = 0, las variables X e Y son **independientes**. La distribución tiene forma circular vista desde arriba.

```python
# Parámetros: media en (0,0), sin correlación
media = [0, 0]
covarianza = [[1, 0],    # ρ = 0, variables independientes
              [0, 1]]

# Creamos la distribución
distribucion = multivariate_normal(media, covarianza)

# Generamos puntos para graficar
x = np.linspace(-3, 3, 100)
y = np.linspace(-3, 3, 100)
X, Y = np.meshgrid(x, y)
pos = np.dstack((X, Y))
Z = distribucion.pdf(pos)

# Gráfico 3D
fig = plt.figure(figsize=(14, 5))

# Vista 3D
ax1 = fig.add_subplot(121, projection='3d')
ax1.plot_surface(X, Y, Z, cmap='viridis', alpha=0.8)
ax1.set_xlabel('X')
ax1.set_ylabel('Y')
ax1.set_zlabel('Densidad')
ax1.set_title('Distribución Normal Bivariada (ρ = 0)\nVista 3D')

# Vista de contorno (desde arriba)
ax2 = fig.add_subplot(122)
contorno = ax2.contourf(X, Y, Z, levels=20, cmap='viridis')
ax2.set_xlabel('X')
ax2.set_ylabel('Y')
ax2.set_title('Curvas de nivel (ρ = 0)\nVista superior')
plt.colorbar(contorno, ax=ax2)
ax2.axis('equal')

plt.tight_layout()
plt.show()

print("📊 Observación: Las curvas de nivel son circulares, lo que indica independencia entre X e Y")
```

## Ejemplo 2: Correlación positiva (ρ = 0.8)

Cuando ρ > 0, existe una **correlación positiva**: cuando X aumenta, Y tiende a aumentar también. Las curvas de nivel son elipses inclinadas hacia la derecha.

```python
# Parámetros con correlación positiva
media = [0, 0]
rho = 0.8
covarianza = [[1, rho],    # Correlación positiva fuerte
              [rho, 1]]

distribucion_pos = multivariate_normal(media, covarianza)
Z_pos = distribucion_pos.pdf(pos)

# Generamos muestra aleatoria para scatter plot
muestra = distribucion_pos.rvs(500)

fig = plt.figure(figsize=(15, 5))

# Vista 3D
ax1 = fig.add_subplot(131, projection='3d')
ax1.plot_surface(X, Y, Z_pos, cmap='plasma', alpha=0.8)
ax1.set_xlabel('X')
ax1.set_ylabel('Y')
ax1.set_zlabel('Densidad')
ax1.set_title(f'Distribución Normal Bivariada (ρ = {rho})')

# Curvas de nivel
ax2 = fig.add_subplot(132)
contorno = ax2.contourf(X, Y, Z_pos, levels=20, cmap='plasma')
ax2.set_xlabel('X')
ax2.set_ylabel('Y')
ax2.set_title('Curvas de nivel')
plt.colorbar(contorno, ax=ax2)
ax2.axis('equal')

# Scatter plot de muestra
ax3 = fig.add_subplot(133)
ax3.scatter(muestra[:, 0], muestra[:, 1], alpha=0.5, s=10)
ax3.set_xlabel('X')
ax3.set_ylabel('Y')
ax3.set_title(f'Muestra aleatoria (n=500)\nρ = {rho}')
ax3.grid(True, alpha=0.3)
ax3.axis('equal')

plt.tight_layout()
plt.show()

print("📈 Observación: Las elipses están inclinadas positivamente.")
print("   Cuando X aumenta, Y también tiende a aumentar.")
```

## Ejemplo 3: Correlación negativa (ρ = -0.7)

Cuando ρ < 0, existe una **correlación negativa**: cuando X aumenta, Y tiende a disminuir. Las elipses están inclinadas hacia la izquierda.

```python
# Parámetros con correlación negativa
media = [0, 0]
rho = -0.7
covarianza = [[1, rho],    # Correlación negativa
              [rho, 1]]

distribucion_neg = multivariate_normal(media, covarianza)
Z_neg = distribucion_neg.pdf(pos)

# Muestra aleatoria
muestra_neg = distribucion_neg.rvs(500)

fig = plt.figure(figsize=(15, 5))

# Vista 3D
ax1 = fig.add_subplot(131, projection='3d')
ax1.plot_surface(X, Y, Z_neg, cmap='coolwarm', alpha=0.8)
ax1.set_xlabel('X')
ax1.set_ylabel('Y')
ax1.set_zlabel('Densidad')
ax1.set_title(f'Distribución Normal Bivariada (ρ = {rho})')

# Curvas de nivel
ax2 = fig.add_subplot(132)
contorno = ax2.contourf(X, Y, Z_neg, levels=20, cmap='coolwarm')
ax2.set_xlabel('X')
ax2.set_ylabel('Y')
ax2.set_title('Curvas de nivel')
plt.colorbar(contorno, ax=ax2)
ax2.axis('equal')

# Scatter plot
ax3 = fig.add_subplot(133)
ax3.scatter(muestra_neg[:, 0], muestra_neg[:, 1], alpha=0.5, s=10, color='coral')
ax3.set_xlabel('X')
ax3.set_ylabel('Y')
ax3.set_title(f'Muestra aleatoria (n=500)\nρ = {rho}')
ax3.grid(True, alpha=0.3)
ax3.axis('equal')

plt.tight_layout()
plt.show()

print("📉 Observación: Las elipses están inclinadas negativamente.")
print("   Cuando X aumenta, Y tiende a disminuir.")
```

## Ejemplo 4: Comparación de diferentes valores de ρ

Veamos cómo cambia la forma de la distribución al variar el coeficiente de correlación.

```python
# Diferentes valores de correlación
rhos = [-0.9, -0.5, 0, 0.5, 0.9]
media = [0, 0]

fig, axes = plt.subplots(2, 5, figsize=(18, 7))

for i, rho in enumerate(rhos):
    # Definir covarianza
    cov = [[1, rho],
           [rho, 1]]
    
    dist = multivariate_normal(media, cov)
    Z_temp = dist.pdf(pos)
    muestra_temp = dist.rvs(300)
    
    # Curvas de nivel
    ax1 = axes[0, i]
    contorno = ax1.contourf(X, Y, Z_temp, levels=15, cmap='viridis')
    ax1.set_title(f'ρ = {rho}')
    ax1.set_xlabel('X')
    if i == 0:
        ax1.set_ylabel('Y')
    ax1.axis('equal')
    
    # Scatter plot
    ax2 = axes[1, i]
    ax2.scatter(muestra_temp[:, 0], muestra_temp[:, 1], alpha=0.4, s=5)
    ax2.set_xlabel('X')
    if i == 0:
        ax2.set_ylabel('Y')
    ax2.grid(True, alpha=0.3)
    ax2.axis('equal')
    ax2.set_xlim(-3, 3)
    ax2.set_ylim(-3, 3)

plt.tight_layout()
plt.show()

print("🔍 Análisis:")
print("  • ρ = -0.9: Correlación negativa muy fuerte (elipse muy inclinada a la izquierda)")
print("  • ρ = -0.5: Correlación negativa moderada")
print("  • ρ = 0: Sin correlación (forma circular)")
print("  • ρ = 0.5: Correlación positiva moderada")
print("  • ρ = 0.9: Correlación positiva muy fuerte (elipse muy inclinada a la derecha)")
```

## Ejemplo 5: Distribución con medias y desviaciones diferentes

Hasta ahora hemos usado μ₁ = μ₂ = 0 y σ₁ = σ₂ = 1. Veamos qué pasa cuando cambiamos estos valores.

```python
# Parámetros diferentes
media = [2, -1]  # Centro desplazado
sigma_x = 2      # Desviación estándar de X
sigma_y = 0.5    # Desviación estándar de Y
rho = 0.6        # Correlación positiva

# Matriz de covarianza
cov = [[sigma_x**2, rho*sigma_x*sigma_y],
       [rho*sigma_x*sigma_y, sigma_y**2]]

dist_custom = multivariate_normal(media, cov)

# Nuevos rangos para X e Y
x_custom = np.linspace(-3, 7, 100)
y_custom = np.linspace(-3, 1, 100)
X_custom, Y_custom = np.meshgrid(x_custom, y_custom)
pos_custom = np.dstack((X_custom, Y_custom))
Z_custom = dist_custom.pdf(pos_custom)

muestra_custom = dist_custom.rvs(500)

fig = plt.figure(figsize=(15, 5))

# Vista 3D
ax1 = fig.add_subplot(131, projection='3d')
ax1.plot_surface(X_custom, Y_custom, Z_custom, cmap='magma', alpha=0.8)
ax1.set_xlabel('X')
ax1.set_ylabel('Y')
ax1.set_zlabel('Densidad')
ax1.set_title(f'μ = {media}, σₓ = {sigma_x}, σᵧ = {sigma_y}, ρ = {rho}')

# Curvas de nivel
ax2 = fig.add_subplot(132)
contorno = ax2.contourf(X_custom, Y_custom, Z_custom, levels=20, cmap='magma')
ax2.scatter(media[0], media[1], color='red', s=100, marker='x', linewidths=3, label='Media')
ax2.set_xlabel('X')
ax2.set_ylabel('Y')
ax2.set_title('Curvas de nivel')
ax2.legend()
plt.colorbar(contorno, ax=ax2)

# Scatter con estadísticas
ax3 = fig.add_subplot(133)
ax3.scatter(muestra_custom[:, 0], muestra_custom[:, 1], alpha=0.5, s=10, color='purple')
ax3.scatter(media[0], media[1], color='red', s=100, marker='x', linewidths=3)
ax3.set_xlabel('X')
ax3.set_ylabel('Y')
ax3.set_title('Muestra aleatoria (n=500)')
ax3.grid(True, alpha=0.3)

plt.tight_layout()
plt.show()

print(f"📊 Parámetros:")
print(f"   Media de X: {media[0]}, Desviación de X: {sigma_x}")
print(f"   Media de Y: {media[1]}, Desviación de Y: {sigma_y}")
print(f"   Correlación: {rho}")
print(f"\n💡 La distribución está centrada en ({media[0]}, {media[1]})")
print(f"   X tiene mayor variabilidad que Y (σₓ = {sigma_x} > σᵧ = {sigma_y})")
```

## Ejemplo 6: Aplicación práctica - Notas de estudiantes

Imaginemos que queremos modelar la relación entre las notas de dos asignaturas: **Álgebra** y **Geometría**.

Supongamos:
- Nota promedio en Álgebra: 5.5 (escala 1-7)
- Nota promedio en Geometría: 5.8
- Desviación estándar en ambas: 0.8
- Correlación: 0.7 (estudiantes buenos en una tienden a ser buenos en la otra)

```python
# Parámetros basados en notas
media_notas = [5.5, 5.8]  # [Álgebra, Geometría]
sigma_algebra = 0.8
sigma_geometria = 0.8
rho_notas = 0.7

cov_notas = [[sigma_algebra**2, rho_notas*sigma_algebra*sigma_geometria],
             [rho_notas*sigma_algebra*sigma_geometria, sigma_geometria**2]]

dist_notas = multivariate_normal(media_notas, cov_notas)

# Generar muestra de 100 estudiantes
notas_estudiantes = dist_notas.rvs(100)

# Asegurar que las notas estén en el rango 1-7
notas_estudiantes = np.clip(notas_estudiantes, 1, 7)

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# Scatter plot de notas
ax1 = axes[0]
scatter = ax1.scatter(notas_estudiantes[:, 0], notas_estudiantes[:, 1], 
                     alpha=0.6, s=50, c=notas_estudiantes[:, 0]+notas_estudiantes[:, 1], 
                     cmap='RdYlGn', edgecolors='black', linewidth=0.5)
ax1.set_xlabel('Nota Álgebra', fontsize=12)
ax1.set_ylabel('Nota Geometría', fontsize=12)
ax1.set_title('Relación entre notas de Álgebra y Geometría\n(100 estudiantes)', fontsize=13)
ax1.grid(True, alpha=0.3)
ax1.set_xlim(3, 7.5)
ax1.set_ylim(3, 7.5)
ax1.axhline(y=4.0, color='red', linestyle='--', alpha=0.5, label='Nota mínima aprobación')
ax1.axvline(x=4.0, color='red', linestyle='--', alpha=0.5)
plt.colorbar(scatter, ax=ax1, label='Suma de notas')
ax1.legend()

# Histogramas marginales
ax2 = axes[1]
ax2.hist(notas_estudiantes[:, 0], bins=15, alpha=0.5, label='Álgebra', color='blue', edgecolor='black')
ax2.hist(notas_estudiantes[:, 1], bins=15, alpha=0.5, label='Geometría', color='green', edgecolor='black')
ax2.set_xlabel('Nota', fontsize=12)
ax2.set_ylabel('Frecuencia', fontsize=12)
ax2.set_title('Distribución de notas por asignatura', fontsize=13)
ax2.legend()
ax2.grid(True, alpha=0.3, axis='y')

plt.tight_layout()
plt.show()

# Estadísticas
print("📚 Estadísticas de las notas simuladas:")
print(f"   Álgebra - Media: {np.mean(notas_estudiantes[:, 0]):.2f}, Desv. Std: {np.std(notas_estudiantes[:, 0]):.2f}")
print(f"   Geometría - Media: {np.mean(notas_estudiantes[:, 1]):.2f}, Desv. Std: {np.std(notas_estudiantes[:, 1]):.2f}")
print(f"   Correlación observada: {np.corrcoef(notas_estudiantes[:, 0], notas_estudiantes[:, 1])[0,1]:.2f}")
print(f"\n✅ Estudiantes que aprueban ambas (nota ≥ 4.0): {np.sum((notas_estudiantes[:, 0] >= 4.0) & (notas_estudiantes[:, 1] >= 4.0))}")
print(f"❌ Estudiantes que reprueban al menos una: {np.sum((notas_estudiantes[:, 0] < 4.0) | (notas_estudiantes[:, 1] < 4.0))}")
```

## Conclusiones

### Puntos clave sobre la distribución normal bivariada:

1. **Coeficiente de correlación (ρ)**:
   - ρ = 0: Variables independientes (curvas circulares)
   - ρ > 0: Correlación positiva (cuando una aumenta, la otra también)
   - ρ < 0: Correlación negativa (cuando una aumenta, la otra disminuye)
   - |ρ| cercano a 1: Correlación muy fuerte

2. **Forma de la distribución**:
   - Las curvas de nivel son elipses
   - La orientación de las elipses depende del signo de ρ
   - El ancho de las elipses depende de las desviaciones estándar

3. **Aplicaciones en educación matemática**:
   - Modelar relaciones entre variables (como notas de asignaturas)
   - Predecir una variable conociendo la otra
   - Analizar la dependencia entre variables
